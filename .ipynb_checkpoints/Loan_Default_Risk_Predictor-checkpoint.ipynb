#**Loan Default Risk Predictor**

## Project Overview

The **Loan Default Risk Predictor** is a machine learning project that predicts the likelihood of a borrower defaulting on a loan using historical loan application data. The project is designed to support micro-lenders and POS agents by providing a quick assessment of a borrower's repayment risk, enabling more informed lending decisions.

The model accepts borrower information through either a CSV file or manual input and outputs a default risk score along with the key factors influencing the prediction. Its performance is evaluated using standard classification metrics, including a confusion matrix, to assess its predictive accuracy and reliability.

The project is implemented in **Python** using **pandas** for data preprocessing and **scikit-learn** for model development and evaluation within **Google Colab**. Deliverables include a Jupyter notebook containing the complete machine learning workflow, a trained model, evaluation results, a comprehensive README, and a short demonstration video showcasing the system's functionality.

```python
# Calculating the Quantile
Q1 = df_loan.select_dtypes(include="number").quantile(0.25)
Q3 = df_loan.select_dtypes(include="number").quantile(0.75)
Qint = Q3 - Q1
#computing the range outside of which Outliers are defined
lower_bound = Q1 - 1.5*Qint
upper_bound = Q3 + 1.5*Qint
display(lower_bound)
display(upper_bound)
```

```python
#Dispaying outliers lower than the lower bounds
for key in lower_bound_dict.keys():
    if (df_loan[key] < lower_bound_dict[key]).any():
        df_lower_outliers = df_loan[df_loan[key] < lower_bound_dict[key]]
        #printing outliers found

        print("----------------------------------------------------------------------------------------------------------")
        print(f"\n Outlier Values of {key} Lower Than The Lower Bounds")
        print("----------------------------------------------------------------------------------------------------------")
        display(df_lower_outliers)

# Displaying outliers greater than the upper bounds
for key in lower_bound_dict.keys():
    if (df_loan[key] > upper_bound_dict[key]).any():
        df_upper_outliers = df_loan[df_loan[key] > upper_bound_dict[key]]
        #removing outliers with masking/filtering
        df_loan = df_loan[df_loan[key] <= upper_bound_dict[key]]
        #printing outliers found

        print("----------------------------------------------------------------------------------------------------------")
        print(f"\n Outlier Values of {key} Higher Than The Upper Bounds")
        print("----------------------------------------------------------------------------------------------------------")

        display(df_upper_outliers)
```

# 5. Machine Learning Workflow

```python
# Feature Matrix
X = df_loan.drop(columns=["Loan_Status"])
target = "Loan_Status"
# Target vector
y = df_loan[target]
print("X shape:", X.shape)
print("y shape:", y.shape)
```

**Train Validation Test Split**

```python
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    random_state=42
)
# Step 2: Split the remaining 85% into train (70% total) and validation (15% total)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.176,
    random_state=42
)
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_val shape:", X_val.shape)
print("y_val shape:", y_val.shape)
```

**Pre-processing Steps**

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python

categorical_features = ["Gender", "Married", "Education", "Self_Employed", "Property_Area"]
numeric_features = ["Dependents", "ApplicantIncome", "CoapplicantIncome", "LoanAmount", "Loan_Amount_Term", "Credit_History"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

**Model Training using XGBoost**

```python
xgb_model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    random_state=42,
    eval_metric="logloss"
)
```

```python
X_train.info()
```

```python
loan_model_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("loan_model", xgb_model)
    ]
)
```

```python
loan_model_pipeline.fit(X_train, y_train)
```

```python
y_pred_val = loan_model_pipeline.predict(X_val)
y_prob_val = loan_model_pipeline.predict_proba(X_val)[:, 1]
```

```python
y_prob_val
```

```python
y_pred_val = loan_model_pipeline.predict(X_val)
y_prob_val = loan_model_pipeline.predict_proba(X_val)[:, 1]
print("Accuracy:", accuracy_score(y_val, y_pred_val))
print("Precision:", precision_score(y_val, y_pred_val))
print("Recall:", recall_score(y_val, y_pred_val))
print("F1:", f1_score(y_val, y_pred_val))
print("ROC-AUC:", roc_auc_score(y_val, y_prob_val))
```

# 6. Hyperparameter Tuning

```python
import optuna
from optuna.samplers import TPESampler
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold, cross_val_score

def objective(trial):
    # 1. Define the hyperparameter search space
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 200),
        'max_depth': trial.suggest_int('max_depth', 7, 10),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'subsample': trial.suggest_float('subsample', 0.1, 0.7),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.7, 1.0),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 10),
        # Crucial for classification (helps with imbalanced classes)
    }

    # 2. Use the Classifier
    model = xgb.XGBClassifier(**params, random_state=42, n_jobs=1)
    loan_model_pipeline_trial = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("loan_model", model)
    ]
)
    # 3. Use Stratified K-Fold to keep class balance intact
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    # 4. Evaluate using an appropriate classification metric (e.g., 'roc_auc' or 'f1')
    score = cross_val_score(loan_model_pipeline_trial, X_train, y_train, cv=cv, scoring='roc_auc').mean()
    return score

# 5. Run the search
sampler = TPESampler(seed=42)
study = optuna.create_study(study_name="loan-optima", direction='maximize', sampler=sampler)
study.optimize(objective, n_trials=150)

print("Best Classification Parameters:", study.best_params)
```

```python
study.best_params
```

```python

# Training Model with the best params found after tuning
final_loan_model = xgb.XGBClassifier(**study.best_params, random_state=42)
final_loan_model_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("loan_model", final_loan_model)
    ]
)
final_loan_model_pipeline.fit(X_train, y_train)
```

```python
y_pred = final_loan_model_pipeline.predict(X_val)
y_prob = final_loan_model_pipeline.predict_proba(X_val)[:, 1]
print("Accuracy:", accuracy_score(y_val, y_pred))
print("Precision:", precision_score(y_val, y_pred))
print("Recall:", recall_score(y_val, y_pred))
print("F1:", f1_score(y_val, y_pred))
print("ROC-AUC:", roc_auc_score(y_val, y_prob))
```


# **POS Loan Business Scenario**
**For a Loan business giving a 60 day loan of 398000 Naira with a total interest of 238000 Naira, the total profit on a good loan (True Negative) is 238000 Naira, and the total loss on a bad loan (False Negative) is 398000 Naira**

```python

#Evaluating several probability thresholds

thresholds = np.arange(0.10, 0.61, 0.015)

results = []

for threshold in thresholds:
    y_val_pred = (y_val_prob >= threshold).astype(int)
    from sklearn.metrics import confusion_matrix
    print("For Threshold:", threshold)
    cm = confusion_matrix(y_val, y_val_pred)

    # Extract TN, FP, FN, TP (Order matters)
    TN, FP, FN, TP = cm.ravel()
    # Per unit loan given out
    TN_income = 238000
    FN_loss = 389000
    # We want to maximize (Total TN income - Total FN loss)

    sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title("XGBoost Confusion Matrix")
    plt.show()

    results.append({
        "Threshold": threshold,
        "Precision": precision_score(y_val, y_val_pred, zero_division=0),
        "Recall": recall_score(y_val, y_val_pred, zero_division=0),
        "Net_Profit": TN * TN_income - FN * FN_loss,
        "Net_Income": TN * TN_income,
        "Net_Loss": FN * FN_loss,
        "F1": f1_score(y_val, y_val_pred, zero_division=0),
        "ROC_AUC": roc_auc_score(y_val, y_val_prob)

    })

threshold_results = pd.DataFrame(results).sort_values(by="Net_Profit", ascending=False)

display(threshold_results)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_val, y_val_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("XGBoost Confusion Matrix")
plt.show()
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    threshold_results["Threshold"],
    threshold_results["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    threshold_results["Threshold"],
    threshold_results["Recall"],
    marker="o",
    label="Recall"
)

plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Precision–Recall Trade-off")
plt.legend()
plt.grid()
plt.show()
```
